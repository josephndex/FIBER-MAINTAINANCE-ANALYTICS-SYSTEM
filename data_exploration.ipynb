## 1. Import Required Libraries

## 2. Load Sample Dataset from Database

```python
# Database connection settings
# Using Configuration 1 (Remote Server) - change to 2 for Local Server
config_number = 1

# Get database credentials from environment
DB_NAME = os.getenv(f'DB_NAME_{config_number}', 'fiber_maintainance_department')
DB_HOST = os.getenv(f'DB_HOST_{config_number}', 'localhost')
DB_USER = os.getenv(f'DB_USER_{config_number}', 'root')
DB_PASSWORD = os.getenv(f'DB_PASSWORD_{config_number}', '')

print(f"Connecting to: {DB_HOST}/{DB_NAME}")

# Create database connection using mysql-connector-python
try:
    engine = create_engine(f'mysql+mysqlconnector://{DB_USER}:{DB_PASSWORD}@{DB_HOST}/{DB_NAME}')
    
    # Load a sample of the data (limit to 1000 rows for exploration)
    query = """
    SELECT * FROM engineered_tickets 
    LIMIT 1000
    """
    df = pd.read_sql(query, engine)
    print(f"Successfully loaded {len(df)} records!")
except Exception as e:
    print(f"Error connecting to database: {e}")
    print("Please check your .env file and database credentials")
```

## 6. Check for Missing Values

```python
# Missing values analysis
print("MISSING VALUES ANALYSIS")
print("=" * 50)
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    'Column': missing.index,
    'Missing Count': missing.values,
    'Missing %': missing_pct.values
})

# Show only columns with missing values
missing_with_nulls = missing_df[missing_df['Missing Count'] > 0].sort_values('Missing Count', ascending=False)

if len(missing_with_nulls) > 0:
    print(f"Columns with missing values: {len(missing_with_nulls)}")
    print(missing_with_nulls.to_string(index=False))
else:
    print("No missing values found in the dataset!")
```

## 7. Validate Data Quality

```python
# Check for duplicates
print("DATA QUALITY CHECKS")
print("=" * 50)

# Duplicate rows
if 'INC' in df.columns:
    duplicate_incs = df['INC'].duplicated().sum()
    print(f"Duplicate INC (ticket) values: {duplicate_incs}")
    
duplicate_rows = df.duplicated().sum()
print(f"Fully duplicate rows: {duplicate_rows}")

# Date range
if 'CREATED_TIME' in df.columns:
    print(f"\nDate Range:")
    print(f"  Earliest: {df['CREATED_TIME'].min()}")
    print(f"  Latest: {df['CREATED_TIME'].max()}")

# SLA Breach distribution
if 'EXTERNAL_BREACHED' in df.columns:
    print(f"\nSLA Breach Distribution:")
    breach_counts = df['EXTERNAL_BREACHED'].value_counts()
    for status, count in breach_counts.items():
        print(f"  {status}: {count:,} ({count/len(df)*100:.1f}%)")
```

```python
# Duration/MTTR analysis
if 'DURATION' in df.columns:
    print("DURATION (MTTR) ANALYSIS")
    print("=" * 50)
    
    # Check the format of DURATION column
    print(f"Data type: {df['DURATION'].dtype}")
    print(f"\nSample values:")
    print(df['DURATION'].head(10).tolist())
    
    # Convert duration from HH:MM:SS string to hours
    def duration_to_hours(duration_str):
        try:
            if pd.isna(duration_str):
                return None
            parts = str(duration_str).split(':')
            if len(parts) == 3:
                hours = int(parts[0]) + int(parts[1])/60 + int(parts[2])/3600
                return hours
            return None
        except:
            return None
    
    df['DURATION_HOURS'] = df['DURATION'].apply(duration_to_hours)
    
    print(f"\nConverted to hours:")
    print(f"Mean MTTR: {df['DURATION_HOURS'].mean():.2f} hours")
    print(f"Median MTTR: {df['DURATION_HOURS'].median():.2f} hours")
    print(f"Min MTTR: {df['DURATION_HOURS'].min():.2f} hours")
    print(f"Max MTTR: {df['DURATION_HOURS'].max():.2f} hours")
    print(f"Std Dev: {df['DURATION_HOURS'].std():.2f} hours")
    
    # Check for outliers (> 24h SLA)
    over_sla = (df['DURATION_HOURS'] > 24).sum()
    print(f"\nTickets over 24h SLA: {over_sla:,} ({over_sla/len(df)*100:.1f}%)")
```

## 8. Summary

The `engineered_tickets` table contains **41 columns** with fiber maintenance ticket records. Key columns include:

### Identification & Status
- **INC**: Unique ticket identifier
- **STATUS**: Ticket status
- **SOURCE_FILE**: Data source

### Location & Assignment  
- **REGION**: Geographic region (Mombasa, Nairobi)
- **CLUSTER**: Fiber cluster/zone (currently all null)
- **CITY**: City location
- **SITE_ID/SITE_NAME**: Site information
- **DISPATCHER**: Ticket dispatcher
- **ENGINEER1/ENGINEER2/ENGINEER3**: Assigned engineers

### Service & Issue Details
- **SERVICE**: Type of service (SDC/FTTS, NPS, FTTB, FTTH, etc.)
- **CAUSE**: Root cause category (Core Break, Cable Damage, etc.)
- **CHALLENGE**: Challenges faced during resolution
- **ROOT_CAUSE**: Root cause details
- **CAUSE_OF_FAILURE**: Failure cause
- **SUMMARY**: Ticket summary
- **LINK_DESCRIPTION**: Link description

### Time & SLA Metrics
- **ESCALATED_TIME**: When ticket was escalated
- **DURATION**: Time to resolve (HH:MM:SS format)
- **EXTERNAL_SLA**: External SLA target
- **EXTERNAL_BREACHED**: SLA breach indicator (YES/NO) - 31.7% breach rate
- **EXTERNAL_BREACHED_HOURS**: Hours over SLA if breached
- **INHOUSE_SLA/INHOUSE_BREACHED/INHOUSE_BREACHED_HOURS**: Internal SLA metrics

### Additional Data
- **LATITUDE/LONGITUDE**: GPS coordinates (70% missing)
- **ACCESS_REQUEST_TIME/ACCESS_GRANTED_TIME**: Access timing
- **SECURITY_REQUEST_TIME/SECURITY_ARRIVAL_TIME**: Security timing
- **MATERIAL_USED**: Materials used for repair
- **NOTES**: Additional notes
- **PATROL/CLEARS/UPTIME**: Additional metrics

### Key Statistics
- Mean MTTR: **14.65 hours**
- Median MTTR: **3.73 hours**
- SLA Breach Rate: **31.7%** (317 out of 1000 sampled tickets)
- Tickets over 24h: **11.1%**
